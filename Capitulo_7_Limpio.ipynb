```python
# Paso 1: Instalar dependencias
!pip install -q gradio langchain openai

# Paso 2: Configurar tu clave de OpenAI (si la tienes)
import os
os.environ["OPENAI_API_KEY"] = "sk-..."  # ← Reemplaza con tu clave de OpenAI

# Paso 3: Crear el asistente con memoria
import gradio as gr
from langchain_community.llms import OpenAI
from langchain.chains import ConversationChain
from langchain.memory import ConversationBufferMemory

# Crear modelo de lenguaje
llm = OpenAI(temperature=0.5)

# Memoria para mantener el historial de la conversación
memory = ConversationBufferMemory()
chatbot = ConversationChain(llm=llm, memory=memory)

# Paso 4: Función de interacción
def chat(message, history):
    if "tarea" in message.lower():
        return "Las tareas del capítulo 7 son:\n1. Crear un agente de IA.\n2. Comparar herramientas IA gratuitas.\n3. Crear un sistema RAG con datos de la UNAL."
    elif "resumen" in message.lower():
        return "Puedes pegar un texto aquí y te ayudo a resumirlo."
    elif "curso" in message.lower():
        return "Este es un asistente para ayudarte con temas del curso de Inteligencia Artificial. Pregunta lo que necesites."
    else:
        return chatbot.run(message)

# Paso 5: Interfaz con Gradio
gr.ChatInterface(
    fn=chat,
    title="🤖 Asistente del Curso de IA",
    description="Este asistente puede ayudarte con las tareas y temas del curso de Inteligencia Artificial.",
#    theme="compact"
).launch(share=True)
```

# 3. Desarrollar un sistema RAG sencillo para crear un chat que permita acceder a la información actualizada de la Universidad Nacional.

```python
!pip install -q langchain sentence-transformers faiss-cpu

texto_unal = """
La Universidad Nacional de Colombia es una institución pública con sedes en todo el país.
El próximo proceso de admisión para pregrado será en septiembre de 2025.
La matrícula financiera se paga semestralmente y depende del puntaje socioeconómico del estudiante.
El calendario académico se publica en la página oficial a inicios de cada semestre.
El sistema de créditos académicos permite cursar materias de diferentes facultades.
El carné estudiantil se debe renovar cada año.
"""

with open("informacion_unal.txt", "w", encoding="utf-8") as f:
    f.write(texto_unal)

from langchain.vectorstores import FAISS
from langchain.document_loaders import TextLoader
from langchain.text_splitter import CharacterTextSplitter
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.llms import HuggingFaceHub

loader = TextLoader("informacion_unal.txt")
documents = loader.load()
splitter = CharacterTextSplitter(chunk_size=300, chunk_overlap=30)
docs = splitter.split_documents(documents)

embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
db = FAISS.from_documents(docs, embeddings)

def consulta_rag(pregunta):
    docs_relacionados = db.similarity_search(pregunta, k=2)
    resultado = "\n---\n".join([doc.page_content for doc in docs_relacionados])
    return f"📄 Documentos relevantes:\n{resultado}"

pregunta = "¿Cuándo es el próximo proceso de admisión?"
respuesta = consulta_rag(pregunta)
print("🔎 Pregunta:", pregunta)
print("💬 Respuesta aproximada:\n", respuesta)
```
